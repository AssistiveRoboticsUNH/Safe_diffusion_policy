# Only Fov based

```python
import random
import h5py
import numpy as np
from scipy.spatial import ConvexHull
from scipy.spatial.transform import Rotation as R
import cv2

###############################################################################
# 1) Utility Functions (Quaternion normalization, Spherical/Cartesian Conversions)
###############################################################################
def normalize_quaternion(q):
    norm = np.linalg.norm(q)
    if norm < 1e-6:
        return np.array([0, 0, 0, 1])
    return q / norm

def cart2sph(x, y, z):
    r = np.sqrt(x**2 + y**2 + z**2)
    az = np.arctan2(y, x)
    el = np.arctan2(z, np.sqrt(x**2 + y**2))
    return r, az, el

def sph2cart(r, az, el):
    x = r * np.cos(el) * np.cos(az)
    y = r * np.cos(el) * np.sin(az)
    z = r * np.sin(el)
    return x, y, z

###############################################################################
# 2) Image Generation and Field-of-View (FoV) Helpers
###############################################################################
def generate_image_frame(img_width, img_height, point_coord, dot_dia=10):
    image = np.zeros((img_height, img_width, 3), dtype=np.uint8)
    cv2.circle(image, point_coord, dot_dia//2, (255,255,255), -1)
    return image

def generate_camera_image(camera_pose, camera_fov, point_obj,
                          img_width=640, img_height=480, dot_dia=10):
    cam_pos = np.array(camera_pose[:3])
    cam_quat = camera_pose[3:]
    cam_rot = R.from_quat(cam_quat).as_matrix()
    point_in_cam = np.linalg.inv(cam_rot) @ (np.array(point_obj) - cam_pos)
    h_fov, v_fov = np.radians(camera_fov)
    x_cam, y_cam, z_cam = point_in_cam
    h_angle = np.arctan2(y_cam, x_cam)
    v_angle = np.arctan2(z_cam, x_cam)
    img_x = int((h_angle + h_fov/2) / h_fov * img_width)
    img_y = int((v_angle + v_fov/2) / v_fov * img_height)
    return generate_image_frame(img_width, img_height, (img_x, img_y), dot_dia)

def is_inFOV(camera_pose, camera_fov, point_obj):
    cam_pos = np.array(camera_pose[:3])
    if np.linalg.norm(np.array(point_obj) - cam_pos) < 1e-4:
        return True
    cam_quat = camera_pose[3:]
    cam_rot = R.from_quat(cam_quat).as_matrix()
    point_in_cam = np.linalg.inv(cam_rot) @ (np.array(point_obj) - cam_pos)
    h_fov, v_fov = np.radians(camera_fov)
    x_cam, y_cam, z_cam = point_in_cam
    if x_cam <= 0:
        return False
    h_angle = np.arctan2(y_cam, x_cam)
    v_angle = np.arctan2(z_cam, x_cam)
    return (abs(h_angle) <= h_fov/2) and (abs(v_angle) <= v_fov/2)

###############################################################################
# 3) Minimal 6D Representation Conversion
###############################################################################
def pose7d_to_6d(pose7d):
    pos = pose7d[:3]
    quat = normalize_quaternion(pose7d[3:7])
    rotvec = R.from_quat(quat).as_rotvec()
    return np.hstack([pos, rotvec])

###############################################################################
# 4) Cube Creation (Candidate Viewpoints)
###############################################################################
def make_cube(cx, cy, cz, rotation, side=0.005):
    half = side / 2.0
    corners = np.array([
        [-half, -half, -half],
        [-half, -half,  half],
        [-half,  half, -half],
        [-half,  half,  half],
        [ half, -half, -half],
        [ half, -half,  half],
        [ half,  half, -half],
        [ half,  half,  half]
    ])
    rotated_corners = rotation.apply(corners)
    rotated_corners += np.array([cx, cy, cz])
    return rotated_corners

###############################################################################
# 5) Generate Candidates for Each Demo Using Cube Method and FoV Filtering
###############################################################################
def generate_candidates_for_demo(positions, goal_pose, num_samples=5):
    """
    positions: array of shape (N,3) for the end-effector positions.
    goal_pose: 7D pose to use as the goal (e.g. the pose when the gripper is closed).
    Returns a list of candidate dictionaries.
    """
    n_points = len(positions)
    deg_to_rad = np.pi/180.0
    max_deg = 15.0            # in degrees
    max_side = 0.2
    max_roll = 100 * deg_to_rad
    max_pitch = 100 * deg_to_rad
    max_yaw = 100* deg_to_rad
    lambda_decay = 1.2
    camera_fov = (160,160)
    img_width, img_height, dot_dia = 640, 480, 10

    goal_pos = goal_pose[:3]
    # Here we use a neutral orientation for generating candidates.
    goal_quat = np.array([0,0,0,1])

    dataset = []
    total_candidates = 0
    in_fov_candidates = 0

    for i in range(n_points):
        x, y, z = positions[i]
        dx = x - goal_pos[0]
        dy = y - goal_pos[1]
        dz = z - goal_pos[2]
        r_val, az, el = cart2sph(dx, dy, dz)
        for sample in range(num_samples):
            frac = np.exp(-lambda_decay*(i/(n_points-1))) if n_points > 1 else 1.0
            delta_angle = frac * (max_deg * deg_to_rad)
            az_var = az + random.choice([-1,1]) * delta_angle
            el_var = el + random.choice([-1,1]) * delta_angle
            dx_var, dy_var, dz_var = sph2cart(r_val, az_var, el_var)
            x_new = goal_pos[0] + dx_var
            y_new = goal_pos[1] + dy_var
            z_new = goal_pos[2] + dz_var

            roll_off = frac * max_roll
            pitch_off = frac * max_pitch
            yaw_off = frac * max_yaw
            droll = random.choice([-1,1]) * roll_off
            dpitch = random.choice([-1,1]) * pitch_off
            dyaw = random.choice([-1,1]) * yaw_off

            rot_obj = R.from_euler('zyx', [droll, dpitch, dyaw])
            quat = rot_obj.as_quat()
            quat = normalize_quaternion(quat)

            # Create a small cube around the new position
            candidates_cube = make_cube(x_new, y_new, z_new, rot_obj, side=frac*max_side)
            for corner in candidates_cube:
                total_candidates += 1
                candidate_pose = [corner[0], corner[1], corner[2],
                                  quat[0], quat[1], quat[2], quat[3]]
                in_fov = is_inFOV(candidate_pose, camera_fov, goal_pos)
                if in_fov:
                    in_fov_candidates += 1
                img = generate_camera_image(candidate_pose, camera_fov, goal_pos,
                                            img_width, img_height, dot_dia)
                dataset.append({
                    "trajectory_idx": i,
                    "pose_7d": candidate_pose,
                    "in_fov": in_fov,
                    "image": img
                })

    print(f"Generated {total_candidates} candidate viewpoints; {in_fov_candidates} passed FoV check.")
    filtered = [vp for vp in dataset if vp["in_fov"]]
    return filtered

###############################################################################
# 6) Compute Global Safe Set and Convex Hull in 6D
###############################################################################
def compute_convex_hull_all(all_candidates, all_main_states):
    candidate_poses_6d = np.array([pose7d_to_6d(vp["pose_7d"]) for vp in all_candidates])
    
    main_poses_list = []
    for states in all_main_states:
        # Ensure quaternions are normalized
        states[:, 3:7] = np.array([normalize_quaternion(q) for q in states[:, 3:7]])
        poses_6d = np.array([pose7d_to_6d(p) for p in states])
        main_poses_list.append(poses_6d[::2])
    
    main_poses_6d = np.vstack(main_poses_list)
    all_poses_6d = np.vstack((main_poses_6d, candidate_poses_6d))

    # Add tiny jitter to avoid degeneracies
    all_poses_6d += np.random.uniform(-1e-8, 1e-8, all_poses_6d.shape)

    hull = ConvexHull(all_poses_6d, qhull_options="QJ")
    return all_poses_6d, hull

###############################################################################
# 7) Save Safe Set + Convex Hull and Sample Random 6D Poses for Testing
###############################################################################
def save_and_test_safe_set(all_poses_6d, hull, num_samples=20):
    np.savez("safe_set_6d_simulation.npz",
             safe_set=all_poses_6d,
             hull_equations=hull.equations,
             hull_vertices=hull.vertices)
    print("Safe set and convex hull saved to 'safe_set_6d_simulation.npz'.")

    mins = np.min(all_poses_6d, axis=0)
    maxs = np.max(all_poses_6d, axis=0)
    samples = []
    for _ in range(num_samples):
        sample = np.random.uniform(mins, maxs)
        samples.append(sample)
    samples = np.array(samples)

    tol = 1e-8
    for i, s in enumerate(samples):
        inside = np.all(np.dot(hull.equations[:, :-1], s) + hull.equations[:, -1] <= tol)
        print(f"Sample {i+1} is {'inside' if inside else 'outside'} the safe set.")
    return samples

###############################################################################
# 8) Main Execution for Simulation
###############################################################################
def main_execution():
    file_path = "/Riad/diffusion_policy/full_image_low_lift_ph.hdf5"  # update the path accordingly
    all_candidates = []
    all_main_states = []  # each element: array of 7D poses (T,7)
    
    with h5py.File(file_path, "r") as f:
        demo_keys = list(f["data"].keys())
        print("Available Demos:", demo_keys)

        for demo_name in demo_keys:
            demo_data = f["data"][demo_name]
            obs_group = demo_data["obs"]

            # Extract end-effector position and orientation.
            # The 7D pose is formed by concatenating robot0_eef_pos (3,) and robot0_eef_quat (4,)
            if ("robot0_eef_pos" not in obs_group) or ("robot0_eef_quat" not in obs_group):
                print(f"Demo {demo_name} is missing eef data. Skipping.")
                continue

            eef_pos = obs_group["robot0_eef_pos"][:]    # shape (T,3)
            eef_quat = obs_group["robot0_eef_quat"][:]  # shape (T,4)
            eef_poses_7d = np.hstack([eef_pos, eef_quat])  # shape (T,7)
            all_main_states.append(eef_poses_7d)

            # Determine the goal pose using gripper data.
            # Assuming robot0_gripper_qpos indicates gripper opening/closing (shape: [T,2]).
            if "robot0_gripper_qpos" in obs_group:
                gripper_qpos = obs_group["robot0_gripper_qpos"][:]  # shape (T,2)
                # Here we assume that values below 0.1 mean closed (adjust the threshold as needed).
                # We pick the first index where the gripper is closed.
                closed_indices = np.where(gripper_qpos.mean(axis=1) < 0.1)[0]
                if closed_indices.size > 0:
                    goal_idx = closed_indices[0]
                else:
                    goal_idx = -1  # fallback to last pose
            else:
                goal_idx = -1  # fallback if no gripper data

            goal_pose = eef_poses_7d[goal_idx]
            print(f"Demo {demo_name}: Using pose at index {goal_idx} as goal (gripper closed).")

            positions = eef_poses_7d[:, :3]
            candidates = generate_candidates_for_demo(positions, goal_pose, num_samples=5)
            if not candidates:
                print(f"No candidate points for Demo {demo_name}.")
                continue

            print(f"Demo {demo_name}: {len(candidates)} candidate points passed FoV check.")
            all_candidates.extend(candidates)

    if not all_candidates:
        print("No candidates from any demo. Exiting.")
        return

    print("Safe set generation started for simulation environment...")
    all_poses_6d, hull = compute_convex_hull_all(all_candidates, all_main_states)
    print("Safe set generation ended.")

    save_and_test_safe_set(all_poses_6d, hull, num_samples=20)

if __name__ == "__main__":
    main_execution()
```

```python
import numpy as np
import plotly.graph_objects as go
from scipy.spatial import ConvexHull

def plot_safe_set_3d(safe_set_file="safe_set_6d_panda_hand.npz"):
    """
    Loads a 6D safe set (N,6) from a .npz file, extracts the first 3 dims
    for X, Y, Z, and plots the 3D convex hull using Plotly.
    """
    data = np.load(safe_set_file)
    safe_6d = data["safe_set"]         # shape (N,6)
    hull_equations = data["hull_equations"]  # shape (m,7)
    hull_vertices = data["hull_vertices"]    # not always used for direct plotting

    # Extract the 3D positions
    positions_3d = safe_6d[:, :3]  # shape (N,3)

    # Build a 3D convex hull from these positions
    hull_3d = ConvexHull(positions_3d)

    # Create a Plotly figure
    fig = go.Figure()

    # Plot the 3D hull as a Mesh3d
    fig.add_trace(go.Mesh3d(
        x=positions_3d[:, 0],
        y=positions_3d[:, 1],
        z=positions_3d[:, 2],
        i=hull_3d.simplices[:, 0],
        j=hull_3d.simplices[:, 1],
        k=hull_3d.simplices[:, 2],
        opacity=0.3,
        color='lightblue',
        name='Safe Set Hull'
    ))

    fig.update_layout(
        title="Safe Set (3D Position Projection)",
        scene=dict(
            xaxis_title="X",
            yaxis_title="Y",
            zaxis_title="Z"
        )
    )
    fig.show()

# Usage example:
if __name__ == "__main__":
    plot_safe_set_3d("/Riad/vivid123/safe_set_6d_simulation.npz")
```
